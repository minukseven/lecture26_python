```python
with open(filename, encoding="utf-8") as f_read:
    lines = f_read.readlines()

for line in lines:
    print(line.strip())
```

### 학습내용 되돌아보기(2)

```python
import math
import random

def getarea(r):
    return round(math.pi * r ** 2, 2)

r = round(random.random() * 10, 2)

print(f'원의 반지름: {r}')
print(f'원주율 pi: {round(math.pi, 4)}')
print(f'반지름 {r}인 원의 면적은 {getarea(r)}')
```

### 학습내용 되돌아보기(2)

```python
f = open('pyzenex.txt', 'wt')
f.write('파이썬 철학\n')
f.write('아름다움이 추함 것보다 낫다.\n')
f.write('명시적인 것이 암묵적인 것보다 낫다.\n')
f.write('단순함이 복잡한 것보다 낫다.\n')
f.close()

# ① read() - 전체를 문자열로 읽기
f = open('pyzenex.txt')
print(f.read())
f.close()

# ② readline() - 한 줄씩 반복 읽기
f = open('pyzenex.txt')
for line in f.readlines():
    print(line, end='')
f.close()

# ③ for line in f - 파일 객체 직접 순회
f = open('pyzenex.txt')
for line in f:
    print(line, end='')
f.close()

# ④ readline() - while로 한 줄씩 읽기
f = open('pyzenex.txt')
while True:
    line = f.readline()
    if not line: break
    print(line, end='')
f.close()
```
